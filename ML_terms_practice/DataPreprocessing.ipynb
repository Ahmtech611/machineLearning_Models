```python
# Now lets take another dataset :

Employee_details = {
    "Name" : ["Maria", "Mary", "Mariam", "Muntaha", "Momina"],
    "Age" : [23, 22, None, 20, 25],
    "Salary($)" : [23400, 24500, None, None, 38900],
    "Qualification" : ["BS-IT", "BS-CS", "BS-DS", "BS-ML", "BS-AI"]
}
```

```python
df2 = pd.DataFrame(Employee_details)

df2
```

```python
df2.info()
```

```python
df2.describe()
```

```python
df2.isnull().sum()
```

```python
# Now lets drop null values rows by using dropna method :

print(df2)
df2_drop = df2.dropna()

print(df2_drop)
```

```python
# now lets fill the NULL values in original dataset by using fillna() :

df2['Age'] = df2['Age'].fillna(df2['Age'].mean())

df2['Salary($)'] = df2['Salary($)'].fillna(df2['Salary($)'].mean())

print(df2)
```

```python
df2.to_csv('Data/cleandata_set1.csv', index=False)
```
