```python
# Now I practice on Feauture scaling :

# here I import two important modules from sklearn.preprocessing :
# 1. StandardScalar()
# 2. MinMaxScalar()

from sklearn.preprocessing import StandardScaler, MinMaxScaler
import pandas as pd
```

```python
# Now lets look as syntax of these two feature Scaling Modules :

# 1. First Syntax :
# scaler = StandardScaler()
# X_scaled = scaler.fit_transform()

# 2. Second Syntax :
# scalar = MinMaxScaler()
# X_scaled = scaler.fit_transform()
```

```python
# And now apply these modules on my practice dataset :

Dataset = {
    "StudyHours" : [10, 4, 5, 2, 7],
    "TestScore" : [98, 46, 61, 23, 75]
}

df = pd.DataFrame(Dataset)
df
```

```python
StdScaler = StandardScaler()
df_scaled = StdScaler.fit_transform(df)
```

```python
print(f"StandardScaled dataset :\n {pd.DataFrame(df_scaled, columns = ['StudyHours', 'TestScore'])}")
```

```python
minmax_scaler = MinMaxScaler()
MainMax_scaled = minmax_scaler .fit_transform(df)
```

```python
print(f"MinMaxScaled dataset :\n {pd.DataFrame(MainMax_scaled, columns = ['StudyHours', 'TestScore'])}")
```

```python
# I move on to train_test_split() :

from sklearn.model_selection import train_test_split
```

```python
X_features = df[['StudyHours']]
y_labeled = df[['TestScore']]

X_train, X_test, y_train, y_test = train_test_split(X_features, y_labeled, test_size=0.2, random_state=42)
```

```python
print(f'''X_train : {X_train}\n
 y_train : {y_train}\n
 X_test : {X_test}\n
 y_test : {y_test}''')
```
